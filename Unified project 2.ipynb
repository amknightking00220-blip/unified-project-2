**Feature Engineering**

*Calculate Shipping Lead Time (days)*

```python
df1['Order Date']=pd.to_datetime(df1['Order Date'],format='%d-%m-%Y')
df1['Ship Date']=pd.to_datetime(df1['Ship Date'],format='%d-%m-%Y')
df1.head(3)
```

```python
df1['Shipping Lead Time']=df1['Ship Date']-df1['Order Date']
df1.head(3)
```

```python
print(df1['Region'].unique())
print()
print(df1['State/Province'].unique())
```

*Categorize routes*

```python
df1['Customer Region Route']=df1['Country/Region']+'->'+df1['Region']
df1['Customer State Route']=df1['Region']+'->'+ df1['State/Province']
df1.head(3)
```
